$$Variaciones\ con\ repetición = m^n = 10^{10}$$

**Posibilidades teniendo en cuenta todas las restricciones**

El enunciado indica que existen únicamente dos restricciones obligatorias que invalidarían una solución si no se cumplen:
1. Se debe asignar, al menos, un partido al viernes a las 20h.
2. Se debe asignar, al menos, un partido al lunes a las 20h.

Para calcular matemáticamente las combinaciones válidas, es mucho más sencillo y eficiente utilizar el principio de inclusión-exclusión (calcular los casos totales y restar los casos que rompen las restricciones).

Se definen los siguientes escenarios inválidos:
* Conjunto A (sin viernes): combinaciones donde no se juega ningún partido el viernes. Quedan 9 horarios disponibles para los 10 partidos, por lo que hay $9^{10}$ combinaciones inválidas.
* Conjunto B (sin lunes): combinaciones donde no se juega ningún partido el lunes. Quedan 9 horarios disponibles para los 10 partidos, por lo que hay $9^{10}$ combinaciones inválidas.
* Intersección A y B (ni viernes ni lunes): al restar los conjuntos A y B, se está restando dos veces el escenario donde no hay partidos ni el viernes ni el lunes. Se debe sumar esta intersección una vez para equilibrar matemáticamente la ecuación. Si se quitan ambos días, quedan 8 horarios para 10 partidos, resultando en $8^{10}$ combinaciones.

La fórmula de combinaciones válidas se expresa así:

$$Total\ válidas = Total\ posible - A - B + (A \cap B)$$
$$Total\ válidas = 10^{10} - 9^{10} - 9^{10} + 8^{10}$$

Se realiza el cálculo numérico:
* $10^{10} = 10.000.000.000$
* $9^{10} = 3.486.784.401$
* $8^{10} = 1.073.741.824$

$$Total = 10.000.000.000 - 3.486.784.401 - 3.486.784.401 + 1.073.741.824$$

Total de posibilidades teniendo en cuenta las restricciones: 4.100.173.022 combinaciones válidas.

**(*) ¿Cual es la función objetivo?**

**Según el modelo para el espacio de soluciones (*) ¿cuál es la función objetivo?**

La función objetivo es la expresión matemática que calcula la audiencia total de una jornada en base a los horarios asignados al conjunto de partidos, aplicando los modificadores pertinentes. Se formula de la siguiente manera:

$$Audiencia\ Total = \sum_{i=1}^{10} (Base_i \times Ponderacion\_Horario_i \times Penalizacion\_Coincidencia_i)$$

En esta fórmula, las variables que intervienen se definen del siguiente modo:
* $Base_i$: es la audiencia base del partido $i$ según las categorías de los equipos que se enfrentan.
* $Ponderacion\_Horario_i$: es el coeficiente de reducción de audiencia que se aplica por jugar en un día y hora específicos.
* $Penalizacion\_Coincidencia_i$: es el porcentaje de audiencia restante tras aplicar la penalización correspondiente si existen otros partidos programados exactamente a la misma hora.

---

**(*) ¿Es un problema de maximización o minimización?**

Tal y como se establece en la descripción general del caso práctico, el propósito principal es organizar los horarios de los partidos de liga de cada jornada para conseguir la mayor cantidad de espectadores posibles. Por tanto, es un problema de maximización.


Diseña un algoritmo para resolver el problema por fuerza bruta

```python
def evaluar_audiencia(solucion, partidos, audiencias_base, ponderaciones, penalizaciones):
    audiencia_total = 0

    # se recorren los 10 partidos programados en el calendario propuesto
    for i in range(len(solucion)):
        horario_asignado = solucion[i]

        # 1. obtener audiencia base según las categorías de los equipos
        base = audiencias_base[partidos[i]['categorias']]

        # 2. obtener modificador multiplicador por el día y hora jugados
        modificador = ponderaciones[horario_asignado]

        # 3. calcular penalización contando cuántos partidos coinciden a esa misma hora
        # (se resta 1 para que el partido no se cuente a sí mismo como coincidencia)
        coincidencias = solucion.count(horario_asignado) - 1
        penalizacion = penalizaciones[coincidencias]

        # se aplica la fórmula matemática de la función objetivo
        audiencia_partido = base * modificador * penalizacion

        # se acumula al total de la jornada
        audiencia_total += audiencia_partido

    return audiencia_total

def fuerza_bruta_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones):
    mejor_audiencia = 0
    mejor_solucion = None

    # generación exhaustiva mediante bucles anidados para los 10 partidos
    for p1 in horarios:
        for p2 in horarios:
            for p3 in horarios:
                for p4 in horarios:
                    for p5 in horarios:
                        for p6 in horarios:
                            for p7 in horarios:
                                for p8 in horarios:
                                    for p9 in horarios:
                                        for p10 in horarios:

                                            # se agrupan los horarios asignados en una lista
                                            solucion = [p1, p2, p3, p4, p5, p6, p7, p8, p9, p10]

                                            # comprobar restricciones obligatorias
                                            if 'Viernes_20' not in solucion or 'Lunes_20' not in solucion:
                                                continue

                                            # evaluar la función objetivo
                                            audiencia_actual = evaluar_audiencia(solucion, partidos, audiencias_base, ponderaciones, penalizaciones)

                                            # actualizar si es la mejor hasta ahora
                                            if audiencia_actual > mejor_audiencia:
                                                mejor_audiencia = audiencia_actual
                                                mejor_solucion = solucion

    return mejor_solucion, mejor_audiencia
```

Aunque nos gustaría mostrar la ejecución del algoritmo, debido al gran coste computacional, resulta muy complicado ejecutarlo y mostrar sus resultados porque podría tardar una gran cantidad de tiempo en ejecutarse. Con el fin de estimar el tiempo necesario para ejecutar el algoritmo de fuerza bruta, se ha elaborado el siguiente código. A partir del número total de soluciones posibles y del número de soluciones que cumplen las restricciones especificadas, se mide el coste medio de generar y comprobar una solución, así como el coste medio de evaluar una solución válida. Estos valores permiten conocer el tiempo necesario para recorrer todo el espacio de soluciones.

Como se puede observar, la ejecución completa se estima en unas 2 horas en la máquina utilizada para realizar las pruebas (y el tiempo se multiplica por 10 con cada partido adicional), por lo que ejecutar el algoritmo de fuerza bruta de forma exhaustiva supone un coste computacional elevado.

```python
import time
import random

# Medimos el coste real de una iteración de la fuerza bruta y extrapolamos al total.
# Muestra de soluciones aleatorias (la proporción de válidas es similar a la real, ~41 %).
random.seed(0)
muestra = 100_000
soluciones_muestra = [[random.choice(horarios) for _ in range(10)] for _ in range(muestra)]

# (a) coste de construir la solución y comprobar las restricciones (se hace en las 10^10 iteraciones)
t0 = time.perf_counter()
for s in soluciones_muestra:
    solucion = list(s)
    if 'Viernes_20' not in solucion or 'Lunes_20' not in solucion:
        continue
t_gen = (time.perf_counter() - t0) / muestra

# (b) coste de evaluar_audiencia (solo en las soluciones válidas)
t0 = time.perf_counter()
for s in soluciones_muestra:
    evaluar_audiencia(s, partidos, audiencias_base, ponderaciones, penalizaciones)
t_eval = (time.perf_counter() - t0) / muestra

total_iter = 10**10
validas = 10**10 - 2 * 9**10 + 8**10
horas = (t_gen * total_iter + t_eval * validas) / 3600
print(f"Coste medio por iteración (construir + comprobar): {t_gen * 1e6:.2f} microsegundos")
print(f"Coste medio de evaluar_audiencia:                  {t_eval * 1e6:.2f} microsegundos")
print(f"Soluciones generadas: {total_iter:,} | válidas (evaluadas): {validas:,}".replace(",", "."))
print(f"Tiempo total estimado en esta máquina: {horas:.1f} horas")
print(f"Con un partido más (11 partidos y 10 franjas) sería {horas * 10:.0f} horas")
```

#### Complejidad Fuerza bruta

Notación: $n$ = número de partidos (10) y $m$ = número de franjas horarias (10).

**1) Número de asignaciones generadas.** Los 10 bucles anidados generan todas las variaciones con repetición: $m^{n} = 10^{10}$. El filtro de restricciones (`'Viernes_20' not in solucion ...`) se aplica *después* de generar cada asignación, así que se recorren las $10^{10}$, de las cuales solo $4.100.173.022$ son válidas y llegan a evaluarse.

**2) Coste por asignación.** Hay que tener en cuenta lo que hace realmente el código:
- Construir la lista `solucion` y comprobar las dos restricciones con `in`: $O(n)$.
- `evaluar_audiencia` (solo para las válidas) tiene un bucle de $n$ iteraciones y **en cada una llama a `solucion.count(horario_asignado)`, que recorre toda la lista**, es decir $O(n)$. El coste de la evaluación es por tanto $O(n \cdot n) = O(n^2)$ y **no** $O(n)$.

**3) Complejidad total.**

$$O(n^{2} \cdot m^{n})$$

Con $n = m = 10$ son del orden de $10^{2}\cdot 10^{10} = 10^{12}$ operaciones elementales (un billón).

**Espacio:** $O(n)$ (solo se guarda la asignación actual y la mejor encontrada).

La celda anterior mide el coste real de una iteración y extrapola el tiempo total: del orden de horas en un ordenador moderno (y bastante más en un entorno como Colab, con sesiones limitadas), y se multiplica por 10 con cada partido adicional. Por eso se considera impracticable y se muestra solo como referencia.

#### Mejora del algoritmo (1): heurística voraz (*Greedy*)

La idea consiste en ordenar los partidos de la jornada por audiencia base (según la categoría de los equipos) y, por otra parte, las franjas por su coeficiente de audiencia. Después se asigna al horario con mayor coeficiente el partido con mayor audiencia esperada, y así sucesivamente hasta rellenar la jornada. Cada franja recibe como máximo un partido, de modo que no se producen coincidencias ($Penalizacion\_Coincidencia_i = 1$).

La función objetivo es la misma de antes:

$$Audiencia\ Total = \sum_{i=1}^{10} (Base_i \times Ponderacion\_Horario_i \times Penalizacion\_Coincidencia_i)$$

**Qué se puede afirmar y qué no sobre su optimalidad.**
- Entre las asignaciones *sin coincidencias* (permutaciones de partidos en franjas), emparejar los mayores con los mayores es óptimo por la desigualdad del reordenamiento.
- Sin embargo, **no se puede garantizar que el óptimo global del problema carezca de coincidencias**. El enunciado solo obliga a ocupar el viernes y el lunes; el resto de franjas pueden quedar vacías. Por ello puede compensar dejar vacía una franja de coeficiente bajo (por ejemplo, domingo 12h, 0,45) y juntar dos partidos en una franja de coeficiente alto aun pagando el 25 % de penalización. Más abajo se muestra un contraejemplo concreto y el estudio empírico cuantifica con qué frecuencia ocurre.

Por tanto, el algoritmo voraz es una **heurística**: muy rápida, $O(n \log n)$, pero sin garantía de óptimo.

```python
def greedy_audiencia_base_por_partido(partidos, audiencias_base):
  audiencia_por_partido = []

  for idx, partido in enumerate(partidos):
    categoria_partido = partido["categorias"]
    audiencia_por_partido.append({"idx": idx, "audiencia_base": audiencias_base[categoria_partido]})

  return sorted(
    audiencia_por_partido,
    key = lambda x: x["audiencia_base"],
    reverse=True
  )

def evaluar_audiencia_greedy(solucion, partidos, audiencias_base, ponderaciones):
    audiencia_total = 0

    for idx, horario in enumerate(solucion):
        audiencia_base = audiencias_base[partidos[idx]["categorias"]]
        audiencia_total += audiencia_base * ponderaciones[horario]

    return audiencia_total

def greedy_liga(partidos, audiencias_base, ponderaciones):
  # Se ordenan los partidos por audiencia base.
  partidos_ordenados_audiencia = greedy_audiencia_base_por_partido(partidos, audiencias_base)

  # Para cada horario, se observa sus coeficientes de audiencia
  horarios_ordenados = [
    horario
    for horario, _ in sorted(
        ponderaciones.items(),
        key=lambda x: x[1],
        reverse=True
    )
  ]
  solucion = [None] * len(partidos_ordenados_audiencia)

  # Se asigna al mejor partido al mejor horario disponible.
  for partido, horario in zip(partidos_ordenados_audiencia, horarios_ordenados):
    solucion[partido["idx"]] = horario

  audiencia = evaluar_audiencia_greedy(
      solucion,
      partidos,
      audiencias_base,
      ponderaciones
  )

  return solucion, audiencia


solucion, audiencia_jornada = greedy_liga(partidos, audiencias_base, ponderaciones)
for i, horario in enumerate(solucion):
    print(f"{partidos[i]['partido']} -> {horario}")

print(f"\nAudiencia total para la jornada: {audiencia_jornada:.2f} Mill.")
```

#### Mejora del algoritmo (2): programación dinámica sobre subconjuntos (óptimo exacto)

Como el voraz no garantiza el óptimo, se propone además un algoritmo **exacto** que no es voraz y que mejora de forma drástica la complejidad de la fuerza bruta.

**Observación clave.** La audiencia de una franja solo depende del *conjunto* de partidos que se asignan a ella: si el conjunto $T$ tiene $k$ partidos, la audiencia de esa franja es

$$valor(s, T) = \Big(\sum_{i \in T} Base_i\Big) \times Coef_s \times Penalizacion[k-1]$$

Como la función objetivo es la suma de los valores de las franjas, el problema se puede resolver franja a franja.

**Subproblema.** $f(s, S)$ = máxima audiencia posible al repartir el subconjunto $S$ de partidos entre las $s$ primeras franjas.

**Recurrencia.**

$$f(s, S) = \max_{T \subseteq S} \Big[ f(s-1,\, S \setminus T) + valor(s, T) \Big]$$

con $f(0, \emptyset) = 0$, $f(0, S\neq\emptyset) = -\infty$, y la restricción de que $T \neq \emptyset$ cuando la franja $s$ es el viernes o el lunes. La solución es $f(m, \text{todos})$, y guardando qué $T$ se eligió en cada paso se reconstruye la asignación.

Como $T$ puede ser vacío, el algoritmo considera de forma natural dejar franjas sin usar y colocar varios partidos en una misma franja, es decir, **explora los solapamientos** que el voraz ignora.

```python
def dp_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones):
    # Programación dinámica sobre subconjuntos: devuelve (solucion, audiencia) óptimos.
    n, m = len(partidos), len(horarios)
    bases = [audiencias_base[p["categorias"]] for p in partidos]
    coef = [ponderaciones[h] for h in horarios]
    obligatorias = {horarios.index("Viernes_20"), horarios.index("Lunes_20")}
    TODOS = (1 << n) - 1

    # Precálculo para cada subconjunto (máscara de bits): suma de audiencias base y nº de partidos
    suma = [0.0] * (1 << n)
    tam = [0] * (1 << n)
    for S in range(1, 1 << n):
        bajo = S & -S
        suma[S] = suma[S ^ bajo] + bases[bajo.bit_length() - 1]
        tam[S] = tam[S ^ bajo] + 1

    NO_ALCANZABLE = float("-inf")
    f = [NO_ALCANZABLE] * (1 << n)
    f[0] = 0.0
    previo = []  # previo[s][S] = partidos ya colocados antes de la franja s (para reconstruir)

    for s in range(m):
        g = [NO_ALCANZABLE] * (1 << n)
        prev = [0] * (1 << n)
        for S in range(1 << n):                 # S = partidos ya asignados a las franjas anteriores
            if f[S] == NO_ALCANZABLE:
                continue
            libres = TODOS ^ S
            T = libres                          # T = partidos que se asignan a la franja s
            while True:                         # se recorren todos los subconjuntos T de 'libres'
                if T or s not in obligatorias:  # viernes y lunes no pueden quedar vacíos
                    valor = f[S] + (suma[T] * coef[s] * penalizaciones[tam[T] - 1] if T else 0.0)
                    if valor > g[S | T]:
                        g[S | T] = valor
                        prev[S | T] = S
                if T == 0:
                    break
                T = (T - 1) & libres
        f = g
        previo.append(prev)

    # Reconstrucción de la solución recorriendo las franjas hacia atrás
    solucion = [None] * n
    S = TODOS
    for s in range(m - 1, -1, -1):
        S_prev = previo[s][S]
        T = S ^ S_prev
        for i in range(n):
            if (T >> i) & 1:
                solucion[i] = horarios[s]
        S = S_prev

    return solucion, f[TODOS]


# 1) Resultado sobre la jornada del enunciado
solucion_dp, audiencia_dp = dp_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones)
for i, horario in enumerate(solucion_dp):
    print(f"{partidos[i]['partido']} -> {horario}")
print(f"\nAudiencia total (óptimo exacto): {audiencia_dp:.3f} Mill.")
print(f"Comprobación con la función objetivo original: "
      f"{evaluar_audiencia(solucion_dp, partidos, audiencias_base, ponderaciones, penalizaciones):.3f} Mill.")
```

**Validación de la programación dinámica frente a la fuerza bruta.** Como la fuerza bruta completa es inviable, se comprueba que ambos algoritmos coinciden en instancias reducidas (6 partidos y 6 franjas, incluyendo viernes y lunes), donde la fuerza bruta sí termina ($6^6 = 46.656$ asignaciones).

```python
import itertools
import random

random.seed(11)
horarios_red = ['Viernes_20', 'Sabado_16', 'Sabado_20', 'Domingo_18', 'Domingo_20', 'Lunes_20']
categorias_posibles = list(audiencias_base.keys())

for prueba in range(5):
    partidos_red = [{"partido": f"P{i+1}", "categorias": random.choice(categorias_posibles)} for i in range(6)]

    # Fuerza bruta reducida (mismo criterio de validez: viernes y lunes ocupados)
    mejor_fb = 0
    for sol in itertools.product(horarios_red, repeat=6):
        if 'Viernes_20' in sol and 'Lunes_20' in sol:
            mejor_fb = max(mejor_fb, evaluar_audiencia(list(sol), partidos_red, audiencias_base, ponderaciones, penalizaciones))

    _, mejor_dp = dp_liga(partidos_red, horarios_red, audiencias_base, ponderaciones, penalizaciones)
    print(f"Prueba {prueba+1}: fuerza bruta = {mejor_fb:.4f} | prog. dinámica = {mejor_dp:.4f} | coinciden: {abs(mejor_fb - mejor_dp) < 1e-9}")
```

#### Mejora del algoritmo (3): recocido simulado (metaheurística)

La programación dinámica nos da el óptimo exacto, pero su coste $O(m \cdot 3^{n})$ sigue creciendo de forma exponencial con el número de partidos. Por eso buscamos también un método que pueda escalar a jornadas más grandes y que, a diferencia del voraz, **sí se atreva a probar coincidencias**. Para ello implementamos un **recocido simulado**, que funciona así:

1. **Punto de partida:** no empezamos de cero, sino desde la solución que nos da el algoritmo voraz. Así el recocido arranca desde una solución que ya es buena y solo tiene que mejorarla.

2. **Cómo se mueve:** en cada paso hacemos un pequeño cambio al azar sobre la solución actual. La mitad de las veces cogemos un partido y lo cambiamos a otra franja cualquiera. Así pueden aparecer o desaparecer coincidencias, que es justo lo que el voraz nunca prueba. La otra mitad, intercambiamos los horarios de dos partidos. Si el cambio deja el viernes o el lunes sin partido, lo descartamos directamente.

3. **Cuándo aceptamos el cambio:** si el cambio mejora la audiencia, nos lo quedamos siempre. Si la empeora, a veces también lo aceptamos, con una probabilidad $e^{\Delta / T}$ que es más alta cuanto menor es la pérdida. Puede parecer raro, pero aceptar a veces una solución peor permite llegar más adelante a soluciones mejores.

4. **Enfriamiento:** la temperatura $T$ marca cuánto "se arriesga" el algoritmo. Empieza alta ($T_0 = 0{,}2$), y al principio acepta bastantes empeoramientos y explora mucho. En cada paso la multiplicamos por $\alpha = 0{,}999$, así que va bajando poco a poco. Tras las $K = 5.000$ iteraciones está tan baja que prácticamente solo acepta mejoras y se dedica a afinar la solución.

```python
import math
from collections import Counter

def evaluar_rapido(solucion, bases, ponderaciones, penalizaciones):
    # Misma función objetivo que evaluar_audiencia, pero en O(n): las coincidencias se cuentan una sola vez
    ocupacion = Counter(solucion)
    return sum(b * ponderaciones[h] * penalizaciones[ocupacion[h] - 1] for b, h in zip(bases, solucion))

def recocido_simulado_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones,
                           t_inicial=0.2, enfriamiento=0.999, iteraciones=5000, semilla=None):
    rng = random.Random(semilla)
    n = len(partidos)
    bases = [audiencias_base[p["categorias"]] for p in partidos]
    obligatorias = ("Viernes_20", "Lunes_20")

    # Solución inicial: la del algoritmo voraz
    actual, _ = greedy_liga(partidos, audiencias_base, ponderaciones)
    valor_actual = evaluar_rapido(actual, bases, ponderaciones, penalizaciones)
    mejor, valor_mejor = actual[:], valor_actual
    t = t_inicial

    for _ in range(iteraciones):
        # Generación de un vecino: mover un partido de franja o intercambiar dos partidos
        vecino = actual[:]
        if rng.random() < 0.5:
            vecino[rng.randrange(n)] = rng.choice(horarios)
        else:
            i, j = rng.sample(range(n), 2)
            vecino[i], vecino[j] = vecino[j], vecino[i]

        # Restricción obligatoria: viernes y lunes ocupados
        if any(h not in vecino for h in obligatorias):
            t *= enfriamiento
            continue

        valor_vecino = evaluar_rapido(vecino, bases, ponderaciones, penalizaciones)
        delta = valor_vecino - valor_actual

        # Criterio de Metrópolis: se aceptan siempre las mejoras y a veces los empeoramientos
        if delta >= 0 or rng.random() < math.exp(delta / t):
            actual, valor_actual = vecino, valor_vecino
            if valor_actual > valor_mejor:
                mejor, valor_mejor = actual[:], valor_actual

        t *= enfriamiento

    return mejor, valor_mejor


# 1) Jornada del enunciado
sol_rs, aud_rs = recocido_simulado_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones, semilla=0)
for i, horario in enumerate(sol_rs):
    print(f"{partidos[i]['partido']} -> {horario}")
print(f"\nRecocido simulado: {aud_rs:.4f} Mill. | óptimo (PD): {audiencia_dp:.4f} Mill.")

# 2) Contraejemplo en el que el voraz no es óptimo
sol_rs_ce, aud_rs_ce = recocido_simulado_liga(contraejemplo, horarios, audiencias_base, ponderaciones, penalizaciones, semilla=0)
print(f"\nContraejemplo -> voraz: {aud_voraz:.4f} | recocido simulado: {aud_rs_ce:.4f} | óptimo (PD): {aud_dp:.4f} Mill.")
print("Franjas con coincidencias en la solución del recocido:",
      {h: c for h, c in Counter(sol_rs_ce).items() if c > 1})
```

## Complejidad

Notación: $n$ = número de partidos y $m$ = número de franjas (en este problema $n = m = 10$).

**Algoritmo voraz (heurística).**
- Ordenar los partidos por audiencia base: $O(n \log n)$.
- Ordenar las franjas por coeficiente: $O(m \log m)$.
- Asignar recorriendo ambas listas y evaluar la solución: $O(n)$.

La complejidad total es $O(n \log n + m \log m)$, que para $n = m$ es $O(n \log n)$. Es la más rápida, pero **no garantiza el óptimo** (ver contraejemplo y estudio empírico).

**Programación dinámica sobre subconjuntos (exacta).**
- Precálculo de la suma de audiencias y del tamaño de cada subconjunto de partidos: $O(2^{n})$.
- Para cada franja ($m$) se recorren todos los pares $(S, T)$ con $T \subseteq$ (partidos no incluidos en $S$). El número de pares es $\sum_{S} 2^{\,n-|S|} = 3^{n}$ y cada transición cuesta $O(1)$ gracias al precálculo.
- Reconstrucción de la solución: $O(m \cdot n)$.

$$T(n, m) = O(m \cdot 3^{n}) \qquad\qquad \text{Memoria: } O(m \cdot 2^{n})$$

(la memoria corresponde a los punteros que se guardan para reconstruir la asignación; si solo se quisiera el valor óptimo bastaría $O(2^{n})$).

**Recocido simulado (metaheurística).**
- Solución inicial con el voraz: $O(n \log n)$.
- En cada una de las $K$ iteraciones se genera un vecino (copia de la solución, $O(n)$), se comprueban las franjas obligatorias ($O(n)$) y se evalúa con `evaluar_rapido`, que cuenta las coincidencias en una sola pasada con `Counter`: $O(n)$.

$$T(n, K) = O(n \log n + K \cdot n) = O(K \cdot n) \qquad\qquad \text{Memoria: } O(n + m)$$

Con $K = 5.000$ y $n = 10$ son unas $5 \cdot 10^{4}$ operaciones. El coste ya no es exponencial en $n$, sino que lo fija el número de iteraciones, por lo que escala a jornadas mucho mayores. A cambio, no garantiza el óptimo.

**Comparación con la fuerza bruta** ($O(n^{2} \cdot m^{n})$):

| Algoritmo | Complejidad | Operaciones con $n = m = 10$ | ¿Óptimo? |
| :--- | :---: | :---: | :---: |
| Fuerza bruta | $O(n^{2} m^{n})$ | $\approx 10^{12}$ | Sí |
| Voraz | $O(n \log n)$ | $\approx 10^{2}$ | No garantizado |
| Prog. dinámica | $O(m \cdot 3^{n})$ | $\approx 5{,}9 \cdot 10^{5}$ | Sí |
| Recocido simulado | $O(K \cdot n)$ | $\approx 5 \cdot 10^{4}$ | No garantizado |

La programación dinámica reduce el coste en un factor del orden de $10^{6}$ manteniendo la garantía de optimalidad. Sigue siendo exponencial en $n$, pero el número de partidos por jornada es pequeño y fijo (10), por lo que es perfectamente viable aquí; para valores de $n$ mayores ($3^{21} \approx 10^{10}$) dejaría de serlo y habría que recurrir a metaheurísticas como el recocido simulado.

### Generar calendario completo

El calendario de la liga de fútbol consta de 38 jornadas si contamos con 20 equipos, donde cada equipo juega de local y visitante ante cada equipo. Desde el año 2018 se cuenta con un calendario totalmente asimétrico, es decir los partidos de la primera y segunda vuelta no se juegan en el mismo orden, es decir no hay relación entre ambas vueltas.

Pero para generar un calendario asimétrico se necesita un algoritmo bastante complejo de implementar con algunas restricciones, por lo tanto se propone adoptar el algoritmo simétrico para generar las jornadas de una manera sencilla y evaluar únicamente los algoritmos de optimización implementados anteriormente.

Para implementar el calendario simétrico, se necesita de un algoritmo "circular" conocido como "Round robin" [2], donde cada equipo obtiene un orden y se van emparejando los equipos según las posiciones y en cada jornada se cambia el orden de una manera circular y se vuelven a emparejar los equipos.

Únicamente hacen falta generar las primeras 19 jornadas y las siguientes 19 serán iguales pero invirtiendo el orden de local y visitante.

Posteriormente utilizamos el algoritmo de optimización para asignar horarios, observar la audiencia potencial por jornada y al final se hace el recuento de la audiencia de la liga al completo.

```python
# Round Robin

def generar_jornada_robin(equipos_jornada, invertir_local_visitante = False):
  partidos = []
  partidos_jornada = len(equipos_jornada) // 2

  for i in range(partidos_jornada):
      local = i
      visitante = -(i + 1)

      if invertir_local_visitante:
          local, visitante = visitante, local

      local = equipos_jornada[local]
      visitante = equipos_jornada[visitante]

      partidos.append({
          "partido": f"{local.nombre} - {visitante.nombre}",
          "categorias": f"{local.categoria}_{visitante.categoria}"
      })

  return partidos

def generar_calendario_robin(equipos_calendario):
  # Calcular cantidad de jornadas ida.
  cantidad_equipos = len(equipos_calendario)
  cantidad_de_jornadas = cantidad_equipos - 1
  calendario = [None] * (cantidad_de_jornadas * 2)

  equipos_round_robin = equipos_calendario[:]
  random.shuffle(equipos_round_robin)

  for i in range(cantidad_de_jornadas):
    calendario[i] = generar_jornada_robin(equipos_round_robin)
    calendario[cantidad_de_jornadas + i] = generar_jornada_robin(equipos_round_robin, invertir_local_visitante=True)

    robin_fijo = equipos_round_robin[0:1]
    robin_variable = equipos_round_robin[-1:] + equipos_round_robin[1:-1]
    equipos_round_robin = robin_fijo + robin_variable

  return calendario

# Generamos el calendario
random.seed(42)
calendario_partido = generar_calendario_robin(equipos)
suma_audiencia_total = 0

for idx_jornada, jornada in enumerate(calendario_partido):
  solucion, audiencia_jornada = greedy_liga(jornada, audiencias_base, ponderaciones)
  suma_audiencia_total += audiencia_jornada

  # Se muestra en detalle solo la primera jornada para no alargar la salida
  if idx_jornada == 0:
    print(f"{'=' * 20} Jornada {idx_jornada + 1} {'=' * 20}")
    for i, horario in enumerate(solucion):
      print(f"{jornada[i]['partido']:<35} → {horario}")
    print("-" * 60)
    print(f"Audiencia de la jornada: {audiencia_jornada:.2f} millones de espectadores.")

print("\n" + "=" * 60)
print(f"Audiencia total del calendario: {suma_audiencia_total:.2f} millones de espectadores")
print("=" * 60)
```

### Estudio empírico de las heurísticas (voraz y recocido simulado)

Se pide evaluar cómo de lejos se quedan las heurísticas del óptimo y qué variabilidad tiene esa diferencia. Como disponemos de un algoritmo exacto (programación dinámica, validado contra la fuerza bruta), el **óptimo real** de cada jornada se calcula sin necesidad de simularlo.

Para cada jornada aleatoria se calcula:
- Diferencia absoluta: $\Delta = Optimo - Heuristica$ (millones de espectadores), donde $Heuristica$ es la audiencia obtenida por el voraz o por el recocido simulado.
- Diferencia relativa: $100 \cdot \Delta / Optimo$ (%).

Como la audiencia solo depende de las audiencias base de los 10 partidos (no de los nombres), el óptimo se guarda en una caché indexada por ese multiconjunto de audiencias, lo que permite evaluar muchas jornadas sin repetir cálculos.

```python
import statistics

def perfil(jornada):
    # Clave de caché: multiconjunto de audiencias base de la jornada
    return tuple(sorted(audiencias_base[p["categorias"]] for p in jornada))

def resumen(nombre, valores):
    print(f"{nombre}: media = {statistics.mean(valores):.5f} | desv. típica = {statistics.pstdev(valores):.5f} "
          f"| varianza = {statistics.pvariance(valores):.2e} | máximo = {max(valores):.5f}")

# --- Experimento 1: jornadas aleatorias de la liga (20 equipos, 10 partidos) ---
random.seed(2024)
N = 10_000
cache_optimo = {}
dif_abs, dif_rel = [], []

for _ in range(N):
    jornada = generar_jornada(equipos)
    clave = perfil(jornada)
    if clave not in cache_optimo:
        cache_optimo[clave] = dp_liga(jornada, horarios, audiencias_base, ponderaciones, penalizaciones)[1]
    optimo = cache_optimo[clave]
    _, voraz = greedy_liga(jornada, audiencias_base, ponderaciones)
    dif_abs.append(optimo - voraz)
    dif_rel.append(100 * (optimo - voraz) / optimo)

subopt = [d for d in dif_abs if d > 1e-9]
print(f"Experimento 1: {N} jornadas aleatorias ({len(cache_optimo)} perfiles de audiencia distintos)")
print(f"Jornadas en las que el voraz alcanza el óptimo: {100 * (N - len(subopt)) / N:.2f} %")
resumen("Diferencia absoluta (Mill.)", dif_abs)
resumen("Diferencia relativa (%)     ", dif_rel)
if subopt:
    print(f"Cuando el voraz no es óptimo ({len(subopt)} jornadas): diferencia media = {statistics.mean(subopt):.4f} Mill.")
```

```python
# --- Experimento 2: calendario completo (38 jornadas Round Robin) con todos los algoritmos ---
import time
import pandas as pd

random.seed(2024)
calendario_estudio = generar_calendario_robin(equipos)
algoritmos = ["Voraz", "Recocido simulado", "Programación dinámica"]
audiencia = {a: [] for a in algoritmos}
tiempo = {a: 0.0 for a in algoritmos}

for k, jornada in enumerate(calendario_estudio):
    t0 = time.perf_counter()
    audiencia["Voraz"].append(greedy_liga(jornada, audiencias_base, ponderaciones)[1])
    tiempo["Voraz"] += time.perf_counter() - t0

    t0 = time.perf_counter()
    audiencia["Recocido simulado"].append(
        recocido_simulado_liga(jornada, horarios, audiencias_base, ponderaciones, penalizaciones, semilla=k)[1])
    tiempo["Recocido simulado"] += time.perf_counter() - t0

    # La programación dinámica se ejecuta sin caché para medir su tiempo real
    t0 = time.perf_counter()
    audiencia["Programación dinámica"].append(
        dp_liga(jornada, horarios, audiencias_base, ponderaciones, penalizaciones)[1])
    tiempo["Programación dinámica"] += time.perf_counter() - t0

optimos = audiencia["Programación dinámica"]
total_optimo = sum(optimos)
filas = []
for a in algoritmos:
    difs = [o - v for o, v in zip(optimos, audiencia[a])]
    filas.append({
        "Algoritmo": a,
        "Audiencia total (M)": f"{sum(audiencia[a]):.3f}",
        "Pérdida vs. óptimo (M)": f"{total_optimo - sum(audiencia[a]):.3f}",
        "Pérdida (%)": f"{100 * (total_optimo - sum(audiencia[a])) / total_optimo:.3f}",
        "Jornadas con el óptimo": f"{sum(d < 1e-9 for d in difs)} / {len(difs)}",
        "Desv. típica dif. por jornada (M)": f"{statistics.pstdev(difs):.5f}",
        "Tiempo total": f"{tiempo[a] * 1e3:.1f} ms",
    })
# La fuerza bruta no se ejecuta: su tiempo se extrapola a partir de la estimación por jornada
filas.append({
    "Algoritmo": "Fuerza bruta (estimada)",
    "Audiencia total (M)": f"{total_optimo:.3f}",
    "Pérdida vs. óptimo (M)": "0.000",
    "Pérdida (%)": "0.000",
    "Jornadas con el óptimo": f"{len(optimos)} / {len(optimos)}",
    "Desv. típica dif. por jornada (M)": "0.00000",
    "Tiempo total": f"≈ {horas * len(calendario_estudio) / 24:.1f} días",
})

print(f"\nExperimento 2: calendario completo de {len(calendario_estudio)} jornadas")
display(pd.DataFrame(filas).set_index("Algoritmo"))
```

#### Estudio empírico del recocido simulado

El recocido simulado es estocástico, así que su diferencia con el óptimo tiene dos fuentes de variabilidad: la jornada (los datos de entrada) y la semilla aleatoria (la ejecución). Se estudian ambas por separado:
- **Experimento 3:** 2.000 jornadas aleatorias, una ejecución por jornada, comparando en las mismas jornadas el voraz y el recocido frente al óptimo de la programación dinámica.
- **Experimento 4:** 200 ejecuciones con semillas distintas sobre el contraejemplo (la jornada en la que el voraz falla), para medir la varianza propia del algoritmo.
- **Tiempos:** coste real de cada algoritmo sobre la jornada del enunciado.


```python
import time

# --- Experimento 3: voraz y recocido simulado frente al óptimo en las mismas jornadas ---
random.seed(2025)
N_RS = 2_000
dif_vz, dif_rs, dif_rel_vz, dif_rel_rs = [], [], [], []

for k in range(N_RS):
    jornada = generar_jornada(equipos)
    clave = perfil(jornada)
    if clave not in cache_optimo:
        cache_optimo[clave] = dp_liga(jornada, horarios, audiencias_base, ponderaciones, penalizaciones)[1]
    optimo = cache_optimo[clave]
    _, vz = greedy_liga(jornada, audiencias_base, ponderaciones)
    _, rs = recocido_simulado_liga(jornada, horarios, audiencias_base, ponderaciones, penalizaciones, semilla=k)
    dif_vz.append(optimo - vz)
    dif_rs.append(optimo - rs)
    dif_rel_vz.append(100 * (optimo - vz) / optimo)
    dif_rel_rs.append(100 * (optimo - rs) / optimo)

def pct_optimo(difs):
    return 100 * sum(d < 1e-9 for d in difs) / len(difs)

print(f"Experimento 3: {N_RS} jornadas aleatorias")
print(f"Voraz    -> alcanza el óptimo en el {pct_optimo(dif_vz):.2f} % de las jornadas")
resumen("  Diferencia absoluta (Mill.)", dif_vz)
resumen("  Diferencia relativa (%)    ", dif_rel_vz)
print(f"Recocido -> alcanza el óptimo en el {pct_optimo(dif_rs):.2f} % de las jornadas")
resumen("  Diferencia absoluta (Mill.)", dif_rs)
resumen("  Diferencia relativa (%)    ", dif_rel_rs)
```

#### Sensibilidad de la heurística a los parámetros

La heurística voraz asume implícitamente que las coincidencias nunca compensan. Para ver hasta qué punto depende eso de los datos del enunciado, repetimos el estudio con 50 jornadas de audiencias base aleatorias (uniformes entre 0,47 y 2,0 M) con (a) la tabla de penalizaciones del enunciado y (b) una tabla de penalizaciones más suave (5 % por coincidencia, con mínimo del 55 %). En ambos casos se compara también el recocido simulado, que sí explora las coincidencias.


```python
random.seed(7)
penal_suaves = {k: max(0.55, 1 - 0.05 * k) for k in range(10)}
res_enunciado, res_suaves = [], []
rs_enunciado, rs_suaves = [], []

for k in range(50):
    audiencias_rand = {f"P{i}": random.uniform(0.47, 2.0) for i in range(10)}
    partidos_rand = [{"partido": f"P{i}", "categorias": f"P{i}"} for i in range(10)]
    _, voraz = greedy_liga(partidos_rand, audiencias_rand, ponderaciones)
    opt_a = dp_liga(partidos_rand, horarios, audiencias_rand, ponderaciones, penalizaciones)[1]
    opt_b = dp_liga(partidos_rand, horarios, audiencias_rand, ponderaciones, penal_suaves)[1]
    rs_a = recocido_simulado_liga(partidos_rand, horarios, audiencias_rand, ponderaciones, penalizaciones, semilla=k)[1]
    rs_b = recocido_simulado_liga(partidos_rand, horarios, audiencias_rand, ponderaciones, penal_suaves, semilla=k)[1]
    res_enunciado.append(100 * (opt_a - voraz) / opt_a)
    res_suaves.append(100 * (opt_b - voraz) / opt_b)
    rs_enunciado.append(100 * (opt_a - rs_a) / opt_a)
    rs_suaves.append(100 * (opt_b - rs_b) / opt_b)

resumen("Voraz,   penalizaciones del enunciado, dif. relativa (%)", res_enunciado)
resumen("Voraz,   penalizaciones suaves,        dif. relativa (%)", res_suaves)
resumen("Recocido, penalizaciones del enunciado, dif. relativa (%)", rs_enunciado)
resumen("Recocido, penalizaciones suaves,        dif. relativa (%)", rs_suaves)
```
