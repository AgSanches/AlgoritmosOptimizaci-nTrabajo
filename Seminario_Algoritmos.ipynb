```
def evaluar_audiencia(solucion, partidos, audiencias_base, ponderaciones, penalizaciones):
    audiencia_total = 0

    # se recorren los 10 partidos programados en el calendario propuesto
    for i in range(len(solucion)):
        horario_asignado = solucion[i]

        # 1. obtener audiencia base según las categorías de los equipos
        base = audiencias_base[partidos[i]['categorias']]

        # 2. obtener modificador multiplicador por el día y hora jugados
        modificador = ponderaciones[horario_asignado]

        # 3. calcular penalización contando cuántos partidos coinciden a esa misma hora
        # (se resta 1 para que el partido no se cuente a sí mismo como coincidencia)
        coincidencias = solucion.count(horario_asignado) - 1
        penalizacion = penalizaciones[coincidencias]

        # se aplica la fórmula matemática de la función objetivo
        audiencia_partido = base * modificador * penalizacion

        # se acumula al total de la jornada
        audiencia_total += audiencia_partido

    return audiencia_total

def fuerza_bruta_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones):
    mejor_audiencia = 0
    mejor_solucion = None

    # generación exhaustiva mediante bucles anidados para los 10 partidos
    for p1 in horarios:
        for p2 in horarios:
            for p3 in horarios:
                for p4 in horarios:
                    for p5 in horarios:
                        for p6 in horarios:
                            for p7 in horarios:
                                for p8 in horarios:
                                    for p9 in horarios:
                                        for p10 in horarios:

                                            # se agrupan los horarios asignados en una lista
                                            solucion = [p1, p2, p3, p4, p5, p6, p7, p8, p9, p10]

                                            # comprobar restricciones obligatorias
                                            if 'Viernes_20' not in solucion or 'Lunes_20' not in solucion:
                                                continue

                                            # evaluar la función objetivo
                                            audiencia_actual = evaluar_audiencia(solucion, partidos, audiencias_base, ponderaciones, penalizaciones)

                                            # actualizar si es la mejor hasta ahora
                                            if audiencia_actual > mejor_audiencia:
                                                mejor_audiencia = audiencia_actual
                                                mejor_solucion = solucion

    return mejor_solucion, mejor_audiencia
```

#### Complejidad Fuerza bruta:

Al anidar 10 bucles iterando sobre 10 horarios posibles, el número total de combinaciones generadas se multiplica en cada paso, resultando en un coste de $m^{n}$, es decir, $10^{10}$ combinaciones.

Además, por cada solución que resulta ser válida, el algoritmo se ve obligado a recorrer la lista de 10 partidos para calcular la suma de la audiencia y aplicar las penalizaciones, lo que añade un coste de tiempo lineal *O(n)* por cada combinación.

La complejidad teórica total del algoritmo se define así: *O(n⋅$m^{n}$)*.

Aplicado a los datos concretos de este problema, la complejidad dictamina que se tendrían que realizar decenas de miles de millones de operaciones básicas. Al observar la forma del código se evidencia visualmente por qué el tiempo de cómputo para los bucles anidados se vuelve inviable para grandes volúmenes de datos, justificando así la utilización de un algoritmo heurístico optimizado en su lugar.

#### Mejora del algoritmo

Para este algoritmo que mejore al de fuerza bruta, se ha optado por elegir *Greedy* (voraz), la idea consiste en ordenar la jornada dada por número de espectadores base, en función de la categoría de los equipos implicados y por otra parte para cada horario, su coeficiente de audiencia.

Posteriormente se otorga al horario con mayor coeficiente, el partido con el mayor número de espectadores esperados hasta rellenar la jornada al completo.

**Análisis de las penalizaciones por coincidencia:**

La función objetivo que modela el problema es la siguiente:

$$Audiencia\ Total = \sum_{i=1}^{10} (Base_i \times Ponderacion\_Horario_i \times Penalizacion\_Coincidencia_i)$$

Aunque formalmente el espacio de soluciones permite el solapamiento de partidos en una misma franja horaria aceptando una penalización, en este problema particular se demuestra analíticamente que la solución óptima global pertenece exclusivamente al conjunto de soluciones sin coincidencias ($Penalizacion\_Coincidencia_i = 1$).

Esto se debe a la naturaleza y relación de los parámetros estáticos del enunciado:
1. **Severidad de la penalización:** el castigo por introducir una sola coincidencia es extremadamente estricto, reduciendo la audiencia un 25% de golpe (aplicando un factor corrector de 0.75).
2. **Gradualidad de los horarios:** los coeficientes de los horarios descienden de manera muy progresiva a lo largo del fin de semana (de 1.0 a 0.85, 0.80, 0.75, etc.).

Debido a esta estructura de datos, el beneficio potencial de desplazar un partido a una franja horaria superior nunca llega a compensar la pérdida del 25% de la audiencia combinada que sufren los partidos solapados. Para la distribución de audiencias base de nuestra liga (cuyo rango se acota entre 2.0 y 0.47 millones), se verifica consistentemente que:

$$F_{\text{con coincidencias}} < F_{\text{sin coincidencias}}$$

Dado que disponemos de exactamente 10 partidos y 10 franjas horarias asignables, siempre es posible construir una solución biyectiva (1 partido en cada franja) donde cada partido ocupe un horario único y libre de penalizaciones.

Por lo tanto, el diseño de nuestro algoritmo voraz prescinde justificadamente de la exploración de solapamientos horarios. Esto nos permite garantizar la obtención del óptimo global con una complejidad temporal de tan solo **$O(n \log n)$** (determinada por el coste de ordenación), reduciendo drásticamente el tiempo de cómputo frente a la fuerza bruta.

```
def greedy_audiencia_base_por_partido(partidos, audiencias_base):
  audiencia_por_partido = []

  for idx, partido in enumerate(partidos):
    categoria_partido = partido["categorias"]
    audiencia_por_partido.append({"idx": idx, "audiencia_base": audiencias_base[categoria_partido]})

  return sorted(
    audiencia_por_partido,
    key = lambda x: x["audiencia_base"],
    reverse=True
  )

def evaluar_audiencia_greedy(solucion, partidos, audiencias_base, ponderaciones):
    audiencia_total = 0

    for idx, horario in enumerate(solucion):
        audiencia_base = audiencias_base[partidos[idx]["categorias"]]
        audiencia_total += audiencia_base * ponderaciones[horario]

    return audiencia_total

def greedy_liga(partidos, audiencias_base, ponderaciones):
  # Se ordenan los partidos por audiencia base.
  partidos_ordenados_audiencia = greedy_audiencia_base_por_partido(partidos, audiencias_base)

  # Para cada horario, se observa sus coeficientes de audencia
  horarios_ordenados = [
    horario
    for horario, _ in sorted(
        ponderaciones.items(),
        key=lambda x: x[1],
        reverse=True
    )
  ]
  solucion = [None] * len(partidos_ordenados_audiencia)

  # Se asigna al mejor partido al mejor horario disponible.
  for partido, horario in zip(partidos_ordenados_audiencia, horarios_ordenados):
    solucion[partido["idx"]] = horario

  audiencia = evaluar_audiencia_greedy(
      solucion,
      partidos,
      audiencias_base,
      ponderaciones
  )

  return solucion, audiencia


solucion, audiencia_jornada = greedy_liga(partidos, audiencias_base, ponderaciones)
for i, horario in enumerate(solucion):
    print(f"{partidos[i]['partido']} -> {horario}")

print(f"\nAudiencia total para la jornada: {audiencia_jornada:.2f} Mill.")
```

```
# Round Robin

def generar_jornada_robin(equipos_jornada, invertir_local_visitante = False):
  partidos = []
  partidos_jornada = len(equipos_jornada) // 2

  for i in range(partidos_jornada):
      local = i
      visitante = -(i + 1)

      if invertir_local_visitante:
          local, visitante = visitante, local

      local = equipos_jornada[local]
      visitante = equipos_jornada[visitante]

      partidos.append({
          "partido": f"{local.nombre} - {visitante.nombre}",
          "categorias": f"{local.categoria}_{visitante.categoria}"
      })

  return partidos

def generar_calendario_robin(equipos_calendario):
  # Calcular cantidad de jornadas ida.
  cantidad_equipos = len(equipos_calendario)
  cantidad_de_jornadas = cantidad_equipos - 1
  calendario = [None] * (cantidad_de_jornadas * 2)

  equipos_round_robin = equipos_calendario[:]
  random.shuffle(equipos_round_robin)

  for i in range(cantidad_de_jornadas):
    calendario[i] = generar_jornada_robin(equipos_round_robin)
    calendario[cantidad_de_jornadas + i] = generar_jornada_robin(equipos_round_robin, invertir_local_visitante=True)

    robin_fijo = equipos_round_robin[0:1]
    robin_variable = equipos_round_robin[-1:] + equipos_round_robin[1:-1]
    equipos_round_robin = robin_fijo + robin_variable

  return calendario

# Generamos el calendario
calendario_partido = generar_calendario_robin(equipos)
suma_audiencia_total = 0

for idx_jornada, jornada in enumerate(calendario_partido):
  solucion, audiencia_jornada = greedy_liga(jornada, audiencias_base, ponderaciones)
  suma_audiencia_total += audiencia_jornada

  print(f"\n{'=' * 20} Jornada {idx_jornada + 1} {'=' * 20}")
  for i, horario in enumerate(solucion):
    print(f"{jornada[i]['partido']:<35} → {horario}")

  print("-" * 60)
  print(f"Audiencia de la jornada: {audiencia_jornada:.2f} millones de espectadores.\n\n")

print("\n" + "=" * 60)
print(f"Audiencia total del calendario: {suma_audiencia_total:.2f} millones de espectadores")
print("=" * 60)
```
